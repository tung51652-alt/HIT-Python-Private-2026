## 1. Từ dữ liệu và hàm đến đối tượng

Ta có thể dùng dict để lưu thông tin học viên và hàm để tính điểm trung bình.


```python
student_data = {"name": "An", "math_score": 8, "literature_score": 9}

def calculate_average(data):
    return (data["math_score"] + data["literature_score"]) / 2

print(student_data["name"])
print(calculate_average(student_data))
```

Khi một đối tượng có nhiều dữ liệu và hành vi liên quan, class giúp tổ chức chúng trong cùng một nơi. Ta sẽ chuyển ví dụ này sang lớp `Student`.


## 2. Class và object

| Khái niệm | Ý nghĩa | Ví dụ |
|---|---|---|
| Class — lớp | Khuôn mẫu mô tả đối tượng | `Student` |
| Object / instance — đối tượng | Một đối tượng cụ thể được tạo từ lớp | `an`, `binh` |
| Attribute — thuộc tính | Dữ liệu của đối tượng | `an.name` |
| Method — phương thức | Hành vi của đối tượng | `an.introduce()` |

### Khởi tạo với `__init__` và `self`

- `__init__` thiết lập dữ liệu ban đầu khi tạo đối tượng.
- `self` tham chiếu đến đối tượng hiện tại.
- `self.name = name` lưu giá trị tham số `name` vào thuộc tính của đối tượng.


```python
class StudentProfile:
    def __init__(self, name, age):
        self.name = name
        self.age = age

an = StudentProfile("An", 18)
binh = StudentProfile("Bình", 19)
print(an.name, an.age)
print(binh.name, binh.age)

an.age = 20
print(an.age, binh.age)
```

`an` và `binh` là hai đối tượng riêng. Thay đổi tuổi của `an` không làm tuổi của `binh` thay đổi.


## 3. Method: hành vi của đối tượng

Method là hàm được định nghĩa trong class. Với instance method, tham số đầu tiên là `self`; khi gọi qua đối tượng, Python tự truyền đối tượng đó vào.

Ví dụ: `introduce()` giới thiệu học viên, `calculate_average()` tính điểm trung bình.


```python
class Student:
    def __init__(self, name, math_score, literature_score):
        self.name = name
        self.math_score = math_score
        self.literature_score = literature_score

    def introduce(self):
        return f"Tôi tên là {self.name}"

    def calculate_average(self):
        return (self.math_score + self.literature_score) / 2

student_an = Student("An", 8, 9)
student_binh = Student("Bình", 6, 8)
print(student_an.introduce())
print(student_an.calculate_average())
print(student_binh.calculate_average())
```

## 4. Thuộc tính instance và class

- **Thuộc tính instance:** dữ liệu riêng của từng đối tượng, thường được gán qua `self` trong `__init__`.
- **Thuộc tính class:** khai báo ở thân class, dùng chung khi các đối tượng chưa có thuộc tính riêng cùng tên.

Trong ví dụ sau, mỗi thành viên có `name` riêng và cùng sử dụng tên trường `school`.


```python
class ClubMember:
    school = "HIT"

    def __init__(self, name):
        self.name = name

member_a = ClubMember("An")
member_b = ClubMember("Bình")
print(member_a.name, member_a.school)
print(member_b.name, member_b.school)

ClubMember.school = "Python Club"
print(member_a.school, member_b.school)
```

## 5. Đóng gói

Đóng gói là tổ chức dữ liệu và các phương thức liên quan trong một lớp, đồng thời cung cấp cách truy cập và cập nhật dữ liệu theo quy tắc.

Với `BankAccount`, ta cần:

- Lưu chủ tài khoản và số dư không âm.
- Chỉ nhận số tiền nạp/rút là số nguyên dương.
- Không cho rút quá số dư hiện có.
- Đọc số dư qua `balance`, cập nhật qua `deposit()` và `withdraw()`.

### Kiểm tra số tiền

Hàm hỗ trợ dưới đây báo `ValueError` khi dữ liệu không hợp lệ. Ta dùng lại hàm này trong lớp tài khoản.

```python
def validate_money(value, allow_zero=False):
    if not isinstance(value, int) or isinstance(value, bool):
        raise ValueError("Số tiền phải là số nguyên, không nhận bool.")
    if value < 0 or (value == 0 and not allow_zero):
        raise ValueError("Số dư phải không âm; số tiền giao dịch phải lớn hơn 0.")
```

### Lớp BankAccount

`_balance` được quy ước là thuộc tính nội bộ. `@property` cho phép đọc bằng `account.balance`; lớp này không cung cấp setter nên việc cập nhật đi qua các phương thức nạp/rút.

`withdraw()` trả về `True` khi rút thành công, `False` khi không đủ tiền. Các bước kiểm tra được thực hiện trước khi thay đổi số dư.

```python
class BankAccount:
    currency = "VND"

    def __init__(self, owner, balance=0):
        validate_money(balance, allow_zero=True)
        self.owner = owner
        self._balance = balance

    @property
    def balance(self):
        return self._balance

    def deposit(self, amount):
        validate_money(amount)
        self._balance += amount

    def withdraw(self, amount):
        validate_money(amount)
        if amount > self._balance:
            return False
        self._balance -= amount
        return True

    def describe(self):
        return f"{self.owner}: {self.balance:,} {self.currency}"
```

### Ví dụ sử dụng

Lan và Minh có hai tài khoản riêng. Quan sát số dư sau mỗi thao tác nạp/rút.

```python
lan = BankAccount("Lan", 100_000)
minh = BankAccount("Minh")

lan.deposit(50_000)
print(lan.withdraw(30_000))
print(lan.withdraw(200_000))
print(lan.describe())
print(minh.describe())
```

Dấu `_` là quy ước sử dụng nội bộ, không ngăn tuyệt đối truy cập từ bên ngoài. Khi dùng lớp, hãy cập nhật số dư qua `deposit()` và `withdraw()` để các quy tắc được kiểm tra.

Với dữ liệu không hợp lệ, có thể dùng `try/except ValueError` ở nơi gọi để xử lý thông báo:

```python
try:
    lan.deposit(-50_000)
except ValueError as error:
    print(error)
```
